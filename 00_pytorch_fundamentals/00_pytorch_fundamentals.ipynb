* **Machine Learning vs. Deep Learning:** Traditional rule-based programming vs. probabilistic pattern extraction; structured tabular models (XGBoost, Random Forest) vs. deep neural representations on unstructured data (text, vision, audio).

# PyTorch for Deep Learning

* Machine Learning is turning things(data) into numbers and finding patterns in those numbers.
* Deep Learning is subset of Machine Learning which is a subset of Artificial Intelligence.

TRADITIONAL PROGRAMMING: INPUTS -> RULES -> OUTPUT

1. Why use ML and DL?
   - Problems with long list of rules (traditional approach listing and managing all rules is too much impossible in case of complex problems but ML/DL can help) 
   - Continuously Changing environments - DL can adapt to new scenarios
   - Discovering insights within large collections of data - Traditional way is too cumbersome.
2. What DL isnt good for?
   - When we need explainability(the patterns of learning interpreting them)
   - When traditional rule based system is better option
   - When errors are unacceptable(DL models are probabilistic)
   - When we dont have much data.
3. ML vs DL:
   - ML operates on structures data
   - Algos like XGBoost
   - Random Forest, Graient Boostted Models, Naive Bayes, Nearest neighbour, Support Vector Machines etc. (shallow algos)
   - vs
   - DL operates on unstructured data like text, images, audio, videos etc.
   - Uses Neural Networks
   - Neural Networks, Fully Connected NN, Convolutional NN, Recurrent NN, Transformers (deeper algos as many layers)
     * Depending how we represent our problems, many algorithms can be used for both structured data and unstructured data.
4. What is NN:
   - Inputs -> Numberical Encoding(converting data into numbers) -> Learns representation(patterns/features/weights)(passing though a NN) -> Representation outputs(numberical output) -> Output(human understandable output)
5. Anantomy of NN:
   INPUT LAYER(data goes in here) -> HIDDEN LAYER(learns patterns in data) -> OUTPUT LAYER(outputs learned representation or prediction probabilities)
    * Each layer is usually a combination of linear(straight line) and/or non-linear(non-straight line) functions.
   
   *Note:* "Patterns" is an arbitrary term, "embeddings", "weights", "feature representation", "feature vectors" all referring to the same things.

7. Types of Learning:
   a. Supervised Learing(data and labels)

   c. Transfer Learning(transfers patterns from one model to another)
   d. Reinforcement Learning(Rewards and Punishment)

## Matrix Multiplication
(Dot Product)

Rules for performing Matrix Multiplication:
1. The **inner dimension** must match:
   - for compatibility of matrix multiplication of `(a,b) @ (c,d)` *b must be equal to c*
(@ symbol is for matrix multiplication)
2. The resulting matrix has the shape of the **outer dimension**:
   - `(a,b) @ (b,c)` will result in a matrix of `(a,c)`

```python
# Element wise multiplication
print(tensor1, "*", tensor2)
tensor1*tensor2
```

```python
# Matrix Multiplication
torch.matmul(tensor1, tensor2) #tensor2 is transposed (1*4+2*5+3*6)
```

```python
import time
# Time difference between manual matrix multiplication vs using torch,matmul()

# Create 2 large tensors
size = 100000
tensorA = torch.rand(size)
tensorB = torch.rand(size)

#----Method 1: Manual loop----
start = time.time()
value = 0
for i in range(size):
    value += tensorA[i] * tensorB[i]
end = time.time()
print(f"Time spent in manual method is {end-start} seconds")

#----Method 2: torch.dot/torch.matmul----
%timeit torch.dot(tensorA, tensorB)

# torchtime = %timeit -o torch.dot(tensorA, tensorB)
# print(f"Time spent in torch method is {torchtime}")

# %%timeit
# value = 0
# for i in range(len(tensor1)):
#     value += tensor1[i] * tensor2[i]
# print(value)

#The massive time difference is due to 
# 1. Interpreted Overhead
#     - check the type, check the bounds(is the index i is in the array), fetch the next instruction
# 2. Vectorization(SIMD-Single Instruction Multiple Data)
#     - instead of 1+1 then 2+2, it does [1,2,3] + [1,2,3] in a single "blink"
# 3. Memory Locality and Caching
#     - manual loops look around in memory for python ojects which are scattered everywhere, causes "cache miss" so CPU have to wait
#     - while tensor are stored in single block of memory so no delay
```

```python
# Rule1: matrix multiplication inner dimensions must match

# torch.matmul(torch.rand(3,2), torch.rand(3,2))
# this gives "RuntimeError: mat1 and mat2 shapes cannot be multiplied (3x2 and 3x2)""

# Rule2: Result is matrix of outer dimensions
tensor_a = torch.rand(3,2)
tensor_b = torch.rand(2,3)
tensor_res = torch.matmul(tensor_a, tensor_b)
tensor_res, tensor_res.shape 
# 3x3 is the outer dimension
```

## Reshaping, Stacking, Squeezing and Unsqueezing tensors
* Reshaping - reshapes an input tensor to a defined shape(should be compatible to the original size)
* View - Return a view of an input tensor of certain shape but keep the same memory as the original tensor
* Stacking - Combine multiple tensors on top of each other (vstack) or side by side (hstack)
* Squeeze - removes all `1` dimensions from a tensor
* Unsqueeze - add a `1` dimensions from a tensor
* Permute - Return a view of the input with dimensions permuted(swapped) in a certain way

```python
# Create a tensor
x = torch.arange(1,10)
x, x.shape, x.ndim
```

```python
# Reshaping
# Add an extra dimension
# x_reshaped = x.reshape(1,7) 
##It will produce an RuntimeError: shape '[1, 7]' is invalid for input of size 9; as we are trying to squeeze 9 elements(original) to 1x7(7 elements)
## (1,8), (1,10), (2,7) all invalid; only(1,9) and (9,1) valid

x_reshaped = x.reshape(1,9)
print(x_reshaped, x_reshaped.shape)
x_reshaped = x.reshape(9,1)
print(x_reshaped, x_reshaped.shape)
```

### Stack
torch.cat() just concatenate the elements without chaging the dimension of tensor(end-to-end long strip(36 numbers) of paper taped(each paper strip contains 9 numbers))

while,

torch.stack() adds a new dimensions and create a grid(one paper strip(9 numbers) placed on top of another)

- vstack(dim=0)
- hstack(dim=1)

```python
# Stack tensors on top of each other
x_stacked = torch.stack([x, x, x, x], dim = 0) #default dimension value is 0. #((9) to (4x9))
x_stacked
# Stack takes a list of tensors and concatenates a sequence of tensors along a new dimension. All tensors need to be of the same size.
```

```python
x_stacked = torch.stack([x, x, x, x], dim = 1) # ((9) to (9x4))
x_stacked
# On using dim = 2; IndexError: Dimension out of range (expected to be in range of [-2, 1], but got 2) 
# due to compatibility with x's original shape(1,9)
```

Here x dimension is 1, 
so when we stack 4 x's [x,x,x,x] using 
dim=0 (stacking as rows): take 4 as first number to insert a new dimension i.e from [9] convert it to [4x9]
dim=1 (stacking as columns): take 4 as second number to insert a new dimension i.e from [9] convert it to [9x4]

in 9x1 there will be two dimensions so dim will go from 0,1,2 and if dim =0 then final result after stack is (4,9,1), dim=1 then(9,4,1) and dim=2 then (9,1,4); coz dim decides which position 4 to add to to make new dimension 3rd

### Squeeze and Unsqueeze

```python
# Squeeze(removes all single diensions from a target tensor)
x_old = torch.rand(1,3,2,1)
x_new = x_old.squeeze()
print("x_old: ", x_old, x_old.shape, x_old.ndim)
print("x_new: ", x_new, x_new.shape, x_new.ndim)

# Unsqueeze(adds single dimenion to a target tensor at a specific dim(dimension))
x_new2 = x_new.unsqueeze(dim = 0)
print("x_new2: ", x_new2, x_new2.shape, x_new2.ndim)
x_new2 = x_new.unsqueeze(dim = 1)
print("x_new2: ", x_new2, x_new2.shape, x_new2.ndim)
x_new2 = x_new.unsqueeze(dim = 2) #originally 4 dimensions squeezed to 2 then unsqueezed to 3 so dim range is [0,1,2] and 3 onwards give error
print("x_new2: ", x_new2, x_new2.shape, x_new2.ndim)
```

```python
# We can also use ":" to select "all" of a target dimension

# (Get all values of 0th dimension and 0th element from that(dimension 1 that's remaining))
# Keep all rows(dim0), but only take first slice of second dimension(dim1). This collapses the middle dimension, leavind a 2D result
print(x_ind[:,0])

# (Get all values of 0th and 1st dimension but only index 1 of second dimension(innermost dimension))
# Keep everything from the first two dimensions, but only grab the second element of the innermost axis
print(x_ind[:, :, 1])

# (Get all values of the 0 dimension but only the 1 index value of the 1st and 2nd dimension)
print(x_ind[:,1,1])

# (Get index 0 of 0th and 1st dimension and all values of 2nd dimension)
print(x_ind[0,0,:])

# Index on x_ind to return 9
print(x_ind[0,2,2])
# Index on x_ind to return [3,6,9]
print(x_ind[0,:,2])
```

## PyTorch Reproducability
#### (trying to take randomness out of random)

- In short how a Neural Network learns:
  `start with random numbers -> tensor operations -> update random numbers to try to make them better representations of the data -> again -> again -> again ...`

- But we might want some less randomness(like when sharing our work with someone)
- To reduce the randomness in neural networks and PyTorch comes the concept of a **random seed**.
- Essentially what the random seed does is "flavour" the randomness.

- What computer basically does is pseudorandomness or generative randomness but random seed flavours that randomness

```python
# Create two random tensors
random_tensor_A = torch.rand(3, 4)
random_tensor_B = torch.rand(3, 4)
print(random_tensor_A)
print(random_tensor_B)
print(random_tensor_A == random_tensor_B)
```

```python
# Random but reproducable tensors
# Set the random seed(this number can be anything)
# RANDOM_SEED = 42
# torch.manual_seed(RANDOM_SEED)
# random_tensor_C = torch.rand(3,4)
# random_tensor_D = torch.rand(3,4)
# print(random_tensor_C)
# print(random_tensor_D)
# print(random_tensor_C == random_tensor_D)

# this will result in false values all as seed is set on only C and not D (if we use torch.manual_seed(RANDOM_SEED) at top it 
#   might work(only 1st time after restarting notebook but C!=D; actually they arent equal but REPRODUCABILITY is executed) 
#   but we have to use it every time we create a random tensor(block))
# this every time creates the same random numbers from page 42(computers generate random numbers using PESUDO-RANDOM NUMBER GENERATOR(PRNG))

RANDOM_SEED = 42
torch.manual_seed(RANDOM_SEED)
random_tensor_C = torch.rand(3,4)
# We must reset the seed before EVERY random operation if we want those specific operations to result in identical tensors.

#RANDOM_SEED = 77 #it will give all false results
torch.manual_seed(RANDOM_SEED)
random_tensor_D = torch.rand(3,4)
print(random_tensor_C)
print(random_tensor_D)
print(random_tensor_C == random_tensor_D)

# We must reset the seed before EVERY random operation if we want those specific operations to result in identical tensors.
#after random seed if we share our work to someone then the person will get the same result as us.
# seed = 42 means open page 42(it can also be 0,1,7,999, etc anything)
# for GPU usage; torch.cuda.manual_seed_all(RANDOM_SEED) is better than torch.manual_seed(RANDOM_SEED)
```

### Putting tensors (and models) on the GPU
- The reason we want our tensors/models on the GPU is because using a GPU results in faster calculations.

```python
# Create a tensor(By default tensor is on  CPU)
tensor = torch.tensor([1, 2, 3])

# Tensor not on GPU
print(tensor, tensor.device)
```

```python
# Move tensor to GPU(if available) #Tensor.to(target device) is used for the purpose
tensor_on_gpu = tensor.to(device) #device here is the target device(gpu here as defined above)
tensor_on_gpu
# cuda:0 is the index of the GPU we are using(as we're using 1 so index is 0, in case of multiple different index will come)
```

### Moving tensors back to the CPU

```python
# If tensor is on GPU, then it cant transform into NumPy
#tensor_on_gpu.numpy()
# it wil result in "TypeError: can't convert cuda:0 device type tensor to numpy. Use Tensor.cpu() to copy the tensor to host memory first."

# copying tensor to CPU
tensor_back_on_cpu = tensor_on_gpu.cpu().numpy()
tensor_back_on_cpu, tensor_on_gpu
```
